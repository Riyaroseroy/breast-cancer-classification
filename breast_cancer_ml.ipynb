```python
import pandas as pd

df = pd.read_csv("data/breast-cancer copy.csv")

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)
```

```python
# Separate features (X) and target (y)

X = df.drop(columns=["id", "diagnosis"])
y = df["diagnosis"].map({"B": 0, "M": 1})

print("X shape:", X.shape)
print("y shape:", y.shape)

print("\nTarget values:")
print(y.value_counts())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    stratify=y,
    random_state=42
)

print("Training set:", X_train.shape)
print("Testing set:", X_test.shape)

print("\nTraining class distribution:")
print(y_train.value_counts())

print("\nTesting class distribution:")
print(y_test.value_counts())
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline

print("Scaling tools imported successfully!")
```

```python
from sklearn.naive_bayes import GaussianNB

# Create the Bayesian classifier pipeline
bayes_pipeline = Pipeline([
    ("scaler", StandardScaler()),
    ("bayes", GaussianNB())
])

print("Bayesian classifier created successfully!")
```

```python
# Train the Bayesian classifier
bayes_pipeline.fit(X_train, y_train)

print("Bayesian classifier trained successfully!")
```

```python
# Make predictions on the unseen test data
y_pred_bayes = bayes_pipeline.predict(X_test)

print("Predictions generated successfully!")
print("Number of predictions:", len(y_pred_bayes))
```

```python
from sklearn.svm import SVC

# Create the SVM pipeline
svm_pipeline = Pipeline([
    ("scaler", StandardScaler()),
    ("svm", SVC(kernel="rbf", C=1.0, probability=True))
])

print("SVM classifier created successfully!")
```

```python
# Make predictions on the unseen test data
y_pred_svm = svm_pipeline.predict(X_test)

print("Predictions generated successfully!")
print("Number of predictions:", len(y_pred_svm))
```

```python
from sklearn.model_selection import StratifiedKFold, cross_val_score

# Create 5-fold stratified cross-validation
cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

print("5-fold cross-validation created successfully!")
```

```python
# Evaluate Bayesian classifier using 5-fold cross-validation
bayes_cv_scores = cross_val_score(
    bayes_pipeline,
    X_train,
    y_train,
    cv=cv,
    scoring="f1"
)

print("Bayesian F1 scores for each fold:")
print(bayes_cv_scores)

print("\nBayesian Mean F1 Score:", bayes_cv_scores.mean())
print("Bayesian Standard Deviation:", bayes_cv_scores.std())
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

# Calculate metrics for Bayesian classifier
bayes_accuracy = accuracy_score(y_test, y_pred_bayes_final)
bayes_precision = precision_score(y_test, y_pred_bayes_final)
bayes_recall = recall_score(y_test, y_pred_bayes_final)
bayes_f1 = f1_score(y_test, y_pred_bayes_final)

# Calculate metrics for SVM
svm_accuracy = accuracy_score(y_test, y_pred_svm_final)
svm_precision = precision_score(y_test, y_pred_svm_final)
svm_recall = recall_score(y_test, y_pred_svm_final)
svm_f1 = f1_score(y_test, y_pred_svm_final)

print("===== FINAL TEST SET RESULTS =====\n")

print("Bayesian Classifier (GaussianNB)")
print("Accuracy :", round(bayes_accuracy, 4))
print("Precision:", round(bayes_precision, 4))
print("Recall   :", round(bayes_recall, 4))
print("F1 Score :", round(bayes_f1, 4))

print("\nSVM")
print("Accuracy :", round(svm_accuracy, 4))
print("Precision:", round(svm_precision, 4))
print("Recall   :", round(svm_recall, 4))
print("F1 Score :", round(svm_f1, 4))
```
